```python
import pandas as pd
```

```python
df = pd.read_csv("P05-02.csv")
```

```python
df
```

```python
# 1. Banyak rumah yang mengizinkan membawa hewan peliharaan dengan harga sewa per tahun di bawah 5000
filt = (df["animal"] == "acept") & (df["rent_amount"] < 5000)
df.loc[filt].shape[0]
```

```python
# 2. Data rumah dengan jumlah kamar mandi terbanyak
df.sort_values("bathroom", ascending=0).head(1)
```

```python
# 3. Banyak rumah dengan luas di atas 200 dan pajak bangunan per tahun di bawah 1700
filt = (df["area"] > 200) & (df["property_tax"] < 1700)
df.loc[filt].shape[0]
```

```python
# 4. Banyak rumah dengan biaya asuransi kebakaran per tahun di atas 70 untuk masing-masing kota
filt = (df["fire_insurance"] > 70)
df.loc[filt, "city"].value_counts()
```

```python
# 5. 5 data rumah dengan jumlah lantai terbanyak yang belum dilengkapi furnitur (not furnished)
filt = df["furniture"] == "not furnished"
df.loc[filt].sort_values("floor", ascending=0).head()
```
