```python
import pandas as pd
```

```python
df = pd.read_csv("P05-01.csv")
```

```python
df
```

```python
# 1. Histogram distribusi total transaksi selama 12 bulan terakhir oleh pengguna dengan jenis kelamin perempuan (F),
# dengan bins 20
filt = (df["Gender"] == "F")
df.loc[filt, "Total_Trans_Amt"].plot(kind="hist", bins=20)
```

```python
# 2. Pie chart perbandingan banyaknya masing-masing status perkawinan dari pengguna
# dengan total transaksi selama 12 bulan terakhir di bawah 2000
filt = (df["Total_Trans_Amt"] < 2000)
df.loc[filt, "Marital_Status"].value_counts().plot(kind="pie")
```

```python
# 3. Scatter plot dengan lama penggunaan kartu kredit dalam bulan sebagai x
# dan rata-rata rasio penggunaan kartu kredit sebagai y
# dari pengguna dengan tipe kartu Blue
filt = (df["Card_Category"] == "Blue")
df.loc[filt].plot(kind="scatter", x="Months_on_book", y="Avg_Utilization_Ratio")
```

```python
# 4. Horizontal bar chart menggambarkan jumlah pengguna dari setiap status keaktifan pengguna
df["Attrition_Flag"].value_counts().plot(kind="barh")
```

```python
# 5. Stacked bar chart dari pengguna dengan kategori penghasilan ’Less than $40K’, ’$40K - $60K’, dan ’$60K- $80K’
# dengan status keaktifan pengguna sebagai sumbu x dan kategori penghasilan sebagai stacked y
filt = (df["Income_Category"] == "Less than $40K") | (df["Income_Category"] == "$40K - $60K") | (df["Income_Category"] == "$60K - $80K")
df.loc[filt].groupby(["Attrition_Flag", "Income_Category"]).size().unstack().plot(kind="bar", stacked=True)
```
