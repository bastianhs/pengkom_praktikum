```python
# 1. Histogram distribusi nilai matematika
df["nilai_mat"].plot(kind="hist", bins=[i for i in range(0, 101, 5)], rwidth=0.9, color="orange", title="Distribusi Nilai Matematika")
```

```python
# 2. Diagram batang horizontal banyaknya mahasiswa masing-masing fakultas
df["fakultas"].value_counts().plot(kind="barh", title="Banyak Mahasiswa Masing-Masing Fakultas", color="red")
```

```python
# 3. Diagram pie banyaknya mahasiswa masing-masing fakultas
df["fakultas"].value_counts().plot(kind="pie", title="Banyak Mahasiswa Masing-Masing Fakultas")
```
